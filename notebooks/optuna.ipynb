```python
import optuna
import optuna.visualization as optuna_vis
from optuna.storages import RDBStorage
import os
import plotly

# Define the storage URL and study name
db_file = os.path.abspath(os.path.join(os.getcwd(), '..','optuna_db.sqlite3'))
db_file_path = f"sqlite:///{db_file}" # The URL format is important
study_name = "ESP32_Kamikaze_Fix"

# --- 1. Load the study from the SQLite database ---
try:
    # We use RDBStorage for relational database backends like SQLite
    storage = RDBStorage(db_file_path)
    
    # Load the existing study by name
    study = optuna.load_study(
        study_name=study_name,
        storage=storage,
    )
    
    print(f"Successfully loaded study '{study_name}' with {len(study.trials)} trials.")

    # --- 2. Generate and display the optimization history graph ---
    # The plot is interactive if run in a compatible environment (like Jupyter Notebook, VS Code with extension)
    # or it opens in a web browser.
    fig = optuna_vis.plot_optimization_history(study)
    fig.show()

except Exception as e:
    print(f"Error loading study or plotting graph: {e}")
    print(f"Please ensure the database file '{db_file_path}' exists and the study name '{study_name}' is correct.")
```
